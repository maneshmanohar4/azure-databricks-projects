```python
from pyspark.sql.functions import *
```

```python
person_df=spark.table('maindatabricksdev.testdata.person')
```

```python
bronze_person_df = person_df.select('BusinessEntityID', 'PersonType',
 'NameStyle',
 'Title',
 concat('FirstName' ,'MiddleName' ,'LastName').alias("FullName"),
 'Suffix',
 'EmailPromotion',
 'AdditionalContactInfo',
 'Demographics')
```

```python
bronze_person_df.write.format('delta').saveAsTable('maindatabricksdev.bronze.bronze_person')
```
