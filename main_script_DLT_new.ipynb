```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import expr
from pyspark.sql.types import StructType
```

```python
%sql
CREATE OR REFRESH STREAMING TABLE credit_card_bronze AS
SELECT * FROM cloud_files("abfss://synapse@testexpstorageacc.dfs.core.windows.net/testdata/live/", "csv", map("schema", "CreditCardID INT, CardType STRING, CardNumber BIGINT, ExpMonth INT, ExpYear INT, ModifiedDate TIMESTAMP"))
```

```python
%sql
CREATE OR REFRESH STREAMING TABLE credit_card_silver AS
SELECT t.CreditCardID, t.CardType, t.CardNumber, t.ExpMonth, t.ExpYear, t.ModifiedDate
FROM STREAM(credit_card_raw_live) t
```

```python
%sql
CREATE OR REFRESH MATERIALIZED VIEW credit_card_gold AS
SELECT CreditCardID, CardType, CardNumber, ExpMonth, ExpYear, ModifiedDate
FROM credit_card_silver
```
